```python
import gradio as gr
import PyPDF2

document_text = ""


def load_document(file):
    global document_text
    if file is None:
        document_text = ""
        return "❌ No document uploaded"

    if file.name.lower().endswith(".pdf"):
        reader = PyPDF2.PdfReader(file)
        text = ""
        for page in reader.pages:
            t = page.extract_text()
            if t:
                text += t + "\n"
        document_text = text
    else:
        document_text = file.read().decode("utf-8", errors="ignore")

    if not document_text.strip():
        return "⚠️ No readable text found"
    return "✅ Document uploaded successfully!"


def get_answer_from_document(question):
    lines = [l.strip() for l in document_text.split("\n") if l.strip()]
    q = question.lower()


    priority = ["degree", "qualification", "name", "email", "phone"]


    for line in lines:
        for key in priority:
            if key in q and key in line.lower():
                return line


    keywords = [w for w in q.split() if len(w) > 3]
    for line in lines:
        if any(k in line.lower() for k in keywords):
            return line

    return "Sorry, the exact answer was not found in the document."


def chat_function(user_input, history):
    msg = user_input.lower()

    # Normal conversation
    if msg in ["hi", "hello", "hey"]:
        return "Hello 👋 How can I help you?"
    if "how are you" in msg:
        return "I’m good 😊 Ask me something from the document."
    if "thank" in msg:
        return "You’re welcome 👍"

    # Document-based
    if not document_text:
        return "📄 Please upload a document first."

    answer = get_answer_from_document(user_input)
    if history:
        last_user, _ = history[-1]
        return f"📄 Answer:\n{answer}\n\n(Follow-up from: '{last_user}')"
    return f"📄 Answer:\n{answer}"

# -------- UI --------
with gr.Blocks() as demo:
    gr.Markdown("## 🤖 Milestone 3 – Document Chatbot")

    with gr.Row():
        with gr.Column(scale=1):
            gr.Markdown("### 📂 Upload Document")
            file_input = gr.File(label="Upload PDF / TXT")
            status = gr.Textbox(label="Status", interactive=False)
            file_input.change(load_document, file_input, status)

        with gr.Column(scale=3):
            gr.ChatInterface(
                fn=chat_function,
                title="Ask questions"
            )

demo.launch()
```
